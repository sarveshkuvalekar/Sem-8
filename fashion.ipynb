```python
import pandas as pd
import numpy as np

from google.colab import files

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
```

```python
train_df = pd.read_csv("fashion-mnist_train.csv")
test_df = pd.read_csv("fashion-mnist_test.csv")
```

```python
# First column contains labels
y_train = train_df.iloc[:, 0]
y_test = test_df.iloc[:, 0]

# Remaining columns contain pixel values
X_train = train_df.iloc[:, 1:]
X_test = test_df.iloc[:, 1:]
```

```python
# Convert into NumPy Arrays

X_train = np.array(X_train)
X_test = np.array(X_test)

y_train = np.array(y_train)
y_test = np.array(y_test)
```

```python
# Normalize Pixel Values

X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# Reshape Images

# Fashion MNIST images are 28x28 grayscale images

X_train = X_train.reshape(-1, 28, 28)
X_test = X_test.reshape(-1, 28, 28)
```

```python
# Convert Labels into Categorical Format

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Build Deep Neural Network Model

model = Sequential()

# Flatten Layer
model.add(Flatten(input_shape=(28, 28)))

# Hidden Layers
model.add(Dense(128, activation='relu'))
model.add(Dense(64, activation='relu'))

# Output Layer
model.add(Dense(10, activation='softmax'))
```

```python
# Compile Model

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train Model

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
# Evaluate Model

loss, accuracy = model.evaluate(X_test, y_test)

print("\nTest Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
# Predictions

predictions = model.predict(X_test)

predicted_classes = np.argmax(predictions, axis=1)
actual_classes = np.argmax(y_test, axis=1)
```

```python
# Display Sample Predictions

fashion_labels = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]

for i in range(5):

    print("\nImage:", i+1)

    print("Actual Label:",
          fashion_labels[actual_classes[i]])

    print("Predicted Label:",
          fashion_labels[predicted_classes[i]])
```
